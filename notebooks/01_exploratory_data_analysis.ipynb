# 01 - Exploratory Data Analysis (EDA)
## UCI AI4I 2020 Predictive Maintenance Dataset

This notebook explores the dataset structure, missing values, class distributions, and sensor telemetry correlations.

```python
import os
import pandas as pd
import numpy as np

DATA_PATH = os.path.join("..", "data", "raw", "ai4i2020.csv")
if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH)
    print(f"Dataset loaded successfully. Shape: {df.shape}")
    display(df.head())
else:
    print(f"Dataset not found at {DATA_PATH}. Please follow instructions in data/README.md to download it.")
```

### Target and Feature Distribution
- Check distribution of `Machine failure` (binary target).
- Verify imbalance ratio.
- Inspect sensor features: Air temperature [K], Process temperature [K], Rotational speed [rpm], Torque [Nm], Tool wear [min].

```python
if 'df' in locals():
    print("\n--- Target Distribution ---")
    print(df['Machine failure'].value_counts(normalize=True))
    print("\n--- Summary Statistics ---")
    display(df.describe())
```
